# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:

**Your answer here:**

Animals migrate for a variety of reasons which include: food scarcity/availability, mating, birthing/raising young, 
avoiding unfavorable weather or temperatures, and avoiding predators who may be migrating into a territory. Climate change can affect animal migrations by disrupting the timing of migrations when the migration is triggered by an environmental condition, such as temperature. Shifting of triggers may drive animals to start their migration either early or late and interfere with the animals' safe transit or the conditions at the destination. Climate change can also impact the conditions along a route or at the destination, which may impact how easy it is for the migrating animals to travel through a location, either in terms of weather, presence of predators, or availabliity of food/prey. 

In Colorado, the migrations I am most aware of are those of Canadian Geese. I have observed increasing presence and transit of the geese in the fall and increased residence through the winter followed by their departure in the spring. Where I used to live in California, we had a significant winter presence of Snow Geese from Canada. Similar to the Canadian Geese, they would arrive in the fall, reside for the winter, and depart in the spring. The presence of the Snow Geese was much more noticeable because the town was small and the hospitable areas for them were fewer, resulting in larger groupings of them.
